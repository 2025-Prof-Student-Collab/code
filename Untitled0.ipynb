## CSV 파일 업로드 & 데이터프레임 저장

```python
# 1. CSV 파일 업로드
from google.colab import files
import pandas as pd

print("불러올 CSV 파일 7개를 모두 선택하세요.")
uploaded = files.upload()

# 2. 읽어서 dataframes 딕셔너리에 저장
dataframes = {}
for filename in uploaded.keys():
    dataframes[filename] = pd.read_csv(filename, encoding='utf-8')

# 3. 파일명을 리스트로 저장
file_names = list(dataframes.keys())

# 각 파일 변수로 저장 (순서 기준)
underweight_children = dataframes[file_names[0]]
underweigh_female = dataframes[file_names[1]]
gdp = dataframes[file_names[2]]
health_expenditure = dataframes[file_names[3]]
maternal_mortality = dataframes[file_names[4]]
first_birth = dataframes[file_names[5]]
school_enrollment = dataframes[file_names[6]]

# 4. 불러온 파일 목록 출력
print("\n불러온 파일 목록:")
for name, df in dataframes.items():
    print(f"- {name} (shape: {df.shape})")

# 5. 각 파일 head 출력
for name, df in dataframes.items():
    print(f"\n===== {name} (head) =====")
    display(df.head())
```

## 필요없는 컬럼 제거 + 컬럼명 확인

```python
# 6. 특정 데이터프레임에서 'Country Code', 'Continent' 삭제
specific_dataframes = [
    underweight_children, underweigh_female, gdp,
    health_expenditure, maternal_mortality, first_birth, school_enrollment
]

column_to_drop = ['Country Code', 'Continent']

for df in specific_dataframes:
    for col in column_to_drop:
        if col in df.columns:
            df.drop(col, axis=1, inplace=True)

# 데이터프레임 이름 리스트
df_names = [
    'underweight_children', 'underweigh_female', 'gdp',
    'health_expenditure', 'maternal_mortality', 'first_birth',
    'school_enrollment'
]

# 7. 컬럼명 확인
for name, df in zip(df_names, specific_dataframes):
    print(f"\n=== {name} columns ===")
    print(list(df.columns))
```

## 파일 병합 (Country Name + Year 기준 조인)

```python
# 8. 파일 병합
df = (underweight_children
      .merge(underweigh_female, on=['Country Name', 'Year'], how='inner')
      .merge(gdp, on=['Country Name', 'Year'], how='inner')
      .merge(health_expenditure, on=['Country Name', 'Year'], how='inner')
      .merge(first_birth, on=['Country Name', 'Year'], how='inner')
      .merge(school_enrollment, on=['Country Name', 'Year'], how='inner'))

df
```

## MICE 결측치 대체 + 결측치 라벨링

```python
# 9. MICE 대체 + 결측치 라벨링
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
import pandas as pd

# MICE 적용 대상 수치형 변수
num_cols = [
    'Percentage of Underweight Children',
    'Prevelance of Underweight among Adults',
    'GDP per capita',
    'Govt. Health Expenditure',
    'Mean age of women at first birth',
    'School enrollment rate, secondary, female'
]

# (1) 결측치 라벨링 표 생성
df_missing_flags = pd.DataFrame()
for col in num_cols:
    df_missing_flags[col + '_was_missing'] = df[col].isnull().astype(int)

# (2) IterativeImputer로 MICE 적용
imputer = IterativeImputer(max_iter=20, random_state=0)
df_num_imputed = pd.DataFrame(
    imputer.fit_transform(df[num_cols]),
    columns=num_cols
)

# (3) 문자형 + 수치형 합치기
df_imputed = pd.concat([
    df[['Country Name', 'Year']].reset_index(drop=True),
    df_num_imputed.reset_index(drop=True)
], axis=1)

# (4) 라벨링까지 합친 최종 데이터
df_imputed_with_flags = pd.concat([
    df_imputed.reset_index(drop=True),
    df_missing_flags.reset_index(drop=True)
], axis=1)

# 출력
print("=== 1) 결측치 라벨링 표(df_missing_flags) ===")
display(df_missing_flags.head())

print("=== 2) MICE 적용된 데이터(df_imputed) ===")
display(df_imputed.head())

print("=== 3) 최종 데이터(df_imputed_with_flags) ===")
display(df_imputed_with_flags.head())
```

## EDA 전체 분석

```python
# 10. EDA 수행
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 그래프 해상도 높게 설정
plt.rcParams['figure.dpi'] = 800

df_eda = df_imputed.copy()

num_cols = [
    'Percentage of Underweight Children',
    'Prevelance of Underweight among Adults',
    'GDP per capita',
    'Govt. Health Expenditure',
    'Mean age of women at first birth',
    'School enrollment rate, secondary, female'
]

# === (1) 기본 구조 ===
print("=== Shape ===")
print(df_eda.shape)

print("\n=== Info ===")
print(df_eda.info())

print("\n=== Head ===")
display(df_eda.head())

# === (2) 결측치 ===
print("\n=== Missing Values ===")
print(df_eda.isnull().sum())

# === (3) 기술통계 ===
print("\n=== Describe ===")
display(df_eda[num_cols].describe())

# === (4) 히스토그램 ===
df_eda[num_cols].hist(bins=20, figsize=(14, 10))
plt.suptitle("Distribution of Numerical Variables")
plt.show()

# === (5) 박스플롯 ===
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_eda[num_cols])
plt.title("Boxplot of Numerical Variables (Check Outliers)")
plt.xticks(rotation=45)
plt.show()

# === (6) pairplot ===
sns.pairplot(df_eda[num_cols], diag_kind='hist')
plt.suptitle("Scatter Matrix", y=1.02)
plt.show()

# === (7) 상관관계 히트맵 ===
plt.figure(figsize=(10, 6))
sns.heatmap(df_eda[num_cols].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

# === (8) 연도별 평균 추세 ===
df_year = df_eda.groupby('Year')[num_cols].mean()

df_year.plot(figsize=(12, 6))
plt.title("Yearly Trends (Mean Values)")
plt.xlabel("Year")
plt.ylabel("Mean Value")
plt.legend(loc='best')
plt.show()

# === (9) 국가별 평균 비교(10개만) ===
df_country = df_eda.groupby('Country Name')[num_cols].mean()
df_country.head(10).plot(kind='bar', figsize=(14, 6))
plt.title("Country-Level Comparison (First 10 Countries)")
plt.xticks(rotation=45)
plt.show()
```
